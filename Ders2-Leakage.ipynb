```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_predict, KFold
from sklearn.metrics import accuracy_score

np.random.seed(42)
N = 500
X = np.random.randn(N, 6)
y = ((X[:,0] + 0.5*X[:,1] - 0.3*X[:,2]) > 0).astype(int)

print("X shape:", X.shape, " sınıf dağılımı:", np.bincount(y))
```

```python
model_leak = RandomForestClassifier(n_estimators=200, random_state=0)
model_leak.fit(X, y)
probs_leak = model_leak.predict_proba(X)   # aynı X üzerinde tahmin!
preds_leak = probs_leak.argmax(axis=1)

acc_leak = accuracy_score(y, preds_leak)
print("Leakage'lı 'accuracy':", acc_leak)
print("Ortalama max olasılık (leakage'lı):", probs_leak.max(axis=1).mean())
```

```python
kf = KFold(n_splits=5, shuffle=True, random_state=0)
model_oof = RandomForestClassifier(n_estimators=200, random_state=0)

probs_oof = cross_val_predict(model_oof, X, y, cv=kf, method="predict_proba")
preds_oof = probs_oof.argmax(axis=1)

acc_oof = accuracy_score(y, preds_oof)
print("OOF (gerçekçi) accuracy:", acc_oof)
print("Ortalama max olasılık (OOF):", probs_oof.max(axis=1).mean())
```

```python
import numpy as np
print("Accuracy farkı (leakage - OOF):", round(acc_leak - acc_oof, 4))
print("Olasılık iyimserlik farkı:", round(probs_leak.max(axis=1).mean() - probs_oof.max(axis=1).mean(), 4))
```
