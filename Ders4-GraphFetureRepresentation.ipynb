```python
import numpy as np
import networkx as nx

channels = ["Fp1", "Fp2", "F3", "O1"]
A = np.array([
    [1.00, 0.80, 0.30, 0.05],
    [0.80, 1.00, 0.25, 0.04],
    [0.30, 0.25, 1.00, 0.15],
    [0.05, 0.04, 0.15, 1.00],
])

G = nx.from_numpy_array(A)
G = nx.relabel_nodes(G, {i: ch for i, ch in enumerate(channels)})
G.remove_edges_from(nx.selfloop_edges(G))
```

```python
degrees = dict(G.degree(weight='weight'))
for ch, d in degrees.items():
    print(f"{ch}: weighted degree = {d:.2f}")
```

```python
# NetworkX ağırlıklı clustering coefficient hesaplayabilir;
# önce sadece belirli bir eşiğin üzerindeki kenarları tutan bir graph oluşturalım
G_thresh = nx.Graph()
G_thresh.add_nodes_from(channels)
threshold = 0.1
for i, ci in enumerate(channels):
    for j, cj in enumerate(channels):
        if i < j and A[i,j] > threshold:
            G_thresh.add_edge(ci, cj)

clustering = nx.clustering(G_thresh)
print(clustering)
```

```python
node_features_handcrafted = np.array([
    [degrees[ch], clustering.get(ch, 0.0)] for ch in channels
])
print("Handcrafted node feature shape:", node_features_handcrafted.shape)
print(node_features_handcrafted)
```
